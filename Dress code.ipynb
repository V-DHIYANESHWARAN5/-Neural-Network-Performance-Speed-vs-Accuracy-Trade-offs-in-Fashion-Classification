```python
import tensorflow as tf
from tensorflow import keras
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

# Enable mixed precision for faster computation (if GPU supports it)
try:
    tf.keras.mixed_precision.set_global_policy('mixed_float16')
    print("Mixed precision enabled for faster training")
except:
    print("Mixed precision not available, using default")

# Set random seeds for reproducibility
np.random.seed(42)
tf.random.set_seed(42)

# Load Fashion MNIST dataset
(x_train_full, y_train_full), (x_test, y_test) = keras.datasets.fashion_mnist.load_data()

print(f"Training data shape: {x_train_full.shape}")
print(f"Test data shape: {x_test.shape}")

# Create class names for Fashion MNIST
class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
               'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']

# Efficient preprocessing - keep as 2D for faster computation
x_train_full = x_train_full.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

# Split training data into training and validation sets
x_train, x_val, y_train, y_val = train_test_split(
    x_train_full, y_train_full, test_size=0.1, random_state=42
)

print(f"Training samples: {len(x_train)}")
print(f"Validation samples: {len(x_val)}")
print(f"Test samples: {len(x_test)}")

# Fast and efficient neural network models

# Model 1: Lightweight model with ReLU + Adam
def create_fast_model_relu_adam():
    model = keras.Sequential([
        # Input layer with proper shape
        keras.layers.Input(shape=(28, 28)),
        
        # Flatten layer to convert 28x28 images to 1D array
        keras.layers.Flatten(),
        
        # Optimized hidden layers - smaller but efficient
        keras.layers.Dense(128, activation='relu'),
        keras.layers.BatchNormalization(),
        keras.layers.Dropout(0.2),
        
        keras.layers.Dense(64, activation='relu'),
        keras.layers.BatchNormalization(),
        keras.layers.Dropout(0.2),
        
        # Output layer
        keras.layers.Dense(10, activation='softmax')
    ])
    
    # Optimizer with default settings for speed
    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    
    return model

# Model 2: Efficient model with Sigmoid + SGD
def create_fast_model_sigmoid_sgd():
    model = keras.Sequential([
        keras.layers.Input(shape=(28, 28)),
        keras.layers.Flatten(),
        
        # Efficient architecture
        keras.layers.Dense(128, activation='sigmoid'),
        keras.layers.BatchNormalization(),
        keras.layers.Dropout(0.2),
        
        keras.layers.Dense(64, activation='sigmoid'),
        
        keras.layers.Dense(10, activation='softmax')
    ])
    
    # SGD optimizer with momentum
    model.compile(
        optimizer=keras.optimizers.SGD(learning_rate=0.1, momentum=0.9, nesterov=True),
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    
    return model

# Model 3: Very fast model with mixed activations
def create_fast_model_mixed():
    model = keras.Sequential([
        keras.layers.Input(shape=(28, 28)),
        keras.layers.Flatten(),
        
        # Mixed activations but fewer layers
        keras.layers.Dense(256, activation='relu'),
        keras.layers.Dropout(0.3),
        
        keras.layers.Dense(128, activation='sigmoid'),
        keras.layers.Dropout(0.2),
        
        keras.layers.Dense(10, activation='softmax')
    ])
    
    # Adam optimizer with higher learning rate for faster convergence
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=0.005),
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    
    return model

# Create models
print("\n" + "="*50)
print("Creating Efficient Models")
print("="*50)

model1 = create_fast_model_relu_adam()
model2 = create_fast_model_sigmoid_sgd()
model3 = create_fast_model_mixed()

print("Model 1 Summary:")
model1.summary()

# Simple callbacks for efficiency
callbacks = [
    keras.callbacks.EarlyStopping(
        monitor='val_accuracy',
        patience=5,
        restore_best_weights=True,
        verbose=1
    ),
    keras.callbacks.ReduceLROnPlateau(
        monitor='val_accuracy',
        factor=0.5,
        patience=3,
        min_lr=0.0001,
        verbose=1
    )
]

# Train Model 1
print("\n" + "="*50)
print("Training Model 1: ReLU activation + Adam optimizer")
print("="*50)
import time

start_time = time.time()
history1 = model1.fit(
    x_train, y_train,
    epochs=20,  # Reduced epochs
    batch_size=128,  # Larger batch size for faster training
    validation_data=(x_val, y_val),
    callbacks=callbacks,
    verbose=1
)
time1 = time.time() - start_time
print(f"Model 1 training time: {time1:.2f} seconds")

# Train Model 2
print("\n" + "="*50)
print("Training Model 2: Sigmoid activation + SGD optimizer")
print("="*50)

start_time = time.time()
history2 = model2.fit(
    x_train, y_train,
    epochs=20,
    batch_size=128,
    validation_data=(x_val, y_val),
    callbacks=callbacks,
    verbose=1
)
time2 = time.time() - start_time
print(f"Model 2 training time: {time2:.2f} seconds")

# Train Model 3
print("\n" + "="*50)
print("Training Model 3: Mixed activations + Adam optimizer")
print("="*50)

start_time = time.time()
history3 = model3.fit(
    x_train, y_train,
    epochs=20,
    batch_size=128,
    validation_data=(x_val, y_val),
    callbacks=callbacks,
    verbose=1
)
time3 = time.time() - start_time
print(f"Model 3 training time: {time3:.2f} seconds")

# Evaluate all models
print("\n" + "="*50)
print("Model Evaluation Results")
print("="*50)

test_accuracies = []
test_losses = []
training_times = [time1, time2, time3]

for i, model in enumerate([model1, model2, model3]):
    test_loss, test_acc = model.evaluate(x_test, y_test, verbose=0)
    test_accuracies.append(test_acc)
    test_losses.append(test_loss)
    print(f"Model {i+1}: Test Accuracy = {test_acc:.4f}, Test Loss = {test_loss:.4f}")

# Create comparison DataFrame
model_names = ['ReLU+Adam', 'Sigmoid+SGD', 'Mixed+Adam']
results_df = pd.DataFrame({
    'Model': model_names,
    'Test Accuracy': test_accuracies,
    'Test Loss': test_losses,
    'Training Time (s)': training_times
})

print("\n" + "="*50)
print("Model Comparison Results")
print("="*50)
print(results_df.to_string(index=False))

# Find best model based on accuracy
best_idx = np.argmax(test_accuracies)
best_model = [model1, model2, model3][best_idx]
best_model_name = model_names[best_idx]

print(f"\nBest model: {best_model_name} with {test_accuracies[best_idx]:.4f} accuracy")

# Efficient visualization
def create_efficient_plots(histories, model_names, test_accuracies, training_times):
    fig, axes = plt.subplots(2, 2, figsize=(12, 10))
    
    # Plot 1: Accuracy comparison
    colors = ['blue', 'green', 'red']
    for i, (history, name, color) in enumerate(zip(histories, model_names, colors)):
        axes[0, 0].plot(history.history['accuracy'], color=color, alpha=0.7, label=f'{name} (train)')
        axes[0, 0].plot(history.history['val_accuracy'], color=color, alpha=0.7, linestyle='--', label=f'{name} (val)')
    
    axes[0, 0].set_title('Model Accuracy Comparison')
    axes[0, 0].set_xlabel('Epoch')
    axes[0, 0].set_ylabel('Accuracy')
    axes[0, 0].legend(loc='lower right')
    axes[0, 0].grid(True, alpha=0.3)
    
    # Plot 2: Test accuracy vs training time
    x_pos = np.arange(len(model_names))
    bars = axes[0, 1].bar(x_pos, test_accuracies, color=colors)
    axes[0, 1].set_xlabel('Model')
    axes[0, 1].set_ylabel('Test Accuracy')
    axes[0, 1].set_title('Test Accuracy Comparison')
    axes[0, 1].set_xticks(x_pos)
    axes[0, 1].set_xticklabels(model_names)
    
    # Add accuracy values on bars
    for bar, acc in zip(bars, test_accuracies):
        height = bar.get_height()
        axes[0, 1].text(bar.get_x() + bar.get_width()/2., height + 0.005,
                       f'{acc:.3f}', ha='center', va='bottom')
    
    # Plot 3: Training time comparison
    axes[1, 0].bar(x_pos, training_times, color=colors)
    axes[1, 0].set_xlabel('Model')
    axes[1, 0].set_ylabel('Training Time (seconds)')
    axes[1, 0].set_title('Training Time Comparison')
    axes[1, 0].set_xticks(x_pos)
    axes[1, 0].set_xticklabels(model_names)
    
    # Plot 4: Loss comparison (only for best model for clarity)
    best_history = histories[best_idx]
    axes[1, 1].plot(best_history.history['loss'], label='Training Loss')
    axes[1, 1].plot(best_history.history['val_loss'], label='Validation Loss')
    axes[1, 1].set_xlabel('Epoch')
    axes[1, 1].set_ylabel('Loss')
    axes[1, 1].set_title(f'Loss for Best Model: {best_model_name}')
    axes[1, 1].legend()
    axes[1, 1].grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()

# Generate plots
create_efficient_plots([history1, history2, history3], model_names, test_accuracies, training_times)

# Efficient prediction visualization
def plot_efficient_predictions(model, images, labels, class_names, n=10):
    # Predict on only n samples
    predictions = model.predict(images[:n], verbose=0)
    predicted_labels = np.argmax(predictions, axis=1)
    
    plt.figure(figsize=(12, 6))
    for i in range(n):
        ax = plt.subplot(2, 5, i+1)
        plt.imshow(images[i], cmap='gray')
        
        true_label = class_names[labels[i]]
        pred_label = class_names[predicted_labels[i]]
        
        if true_label == pred_label:
            color = 'green'
            plt.title(f"{true_label}", color=color)
        else:
            color = 'red'
            plt.title(f"True: {true_label}\nPred: {pred_label}", color=color, fontsize=9)
        
        plt.axis('off')
    
    plt.suptitle(f'Model Predictions (n={n})')
    plt.tight_layout()
    plt.show()

print("\nPredictions from Best Model:")
plot_efficient_predictions(best_model, x_test, y_test, class_names)

# Save models in efficient format
for i, (model, name) in enumerate(zip([model1, model2, model3], model_names)):
    # Remove spaces from name for filename
    filename = f"fashion_mnist_{name.replace('+', '_')}.keras"
    model.save(filename)
    print(f"Model {i+1} ({name}) saved as {filename}")

# Save best model with optimized format
best_model.save("best_fashion_model.keras")
print(f"\nBest model saved as 'best_fashion_model.keras'")

# Additional optimization: Create an even simpler model for comparison
print("\n" + "="*50)
print("Creating Ultra-Fast Simple Model")
print("="*50)

def create_ultra_fast_model():
    model = keras.Sequential([
        keras.layers.Input(shape=(28, 28)),
        keras.layers.Flatten(),
        keras.layers.Dense(64, activation='relu'),
        keras.layers.Dense(10, activation='softmax')
    ])
    
    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    
    return model

ultra_fast_model = create_ultra_fast_model()
ultra_fast_model.summary()

# Train ultra-fast model
start_time = time.time()
ultra_history = ultra_fast_model.fit(
    x_train, y_train,
    epochs=10,  # Very few epochs
    batch_size=256,  # Large batch size
    validation_data=(x_val, y_val),
    verbose=1
)
ultra_time = time.time() - start_time

# Evaluate
ultra_loss, ultra_acc = ultra_fast_model.evaluate(x_test, y_test, verbose=0)
print(f"Ultra-Fast Model: Accuracy = {ultra_acc:.4f}, Loss = {ultra_loss:.4f}, Time = {ultra_time:.2f}s")

# Save ultra-fast model
ultra_fast_model.save("ultra_fast_fashion_model.keras")
print("Ultra-fast model saved as 'ultra_fast_fashion_model.keras'")

print("\n" + "="*50)
print("Performance Summary")
print("="*50)
print(f"Best model accuracy: {test_accuracies[best_idx]:.4f}")
print(f"Ultra-fast model accuracy: {ultra_acc:.4f}")
print(f"Training time improvement: {((time1+time2+time3)/3 - ultra_time)/((time1+time2+time3)/3)*100:.1f}% faster")

# Tips for even faster computation
print("\n" + "="*50)
print("Tips for Even Faster Computation:")
print("="*50)
print("1. Use Google Colab with GPU/TPU")
print("2. Reduce batch size if memory limited")
print("3. Use fewer epochs with early stopping")
print("4. Simplify model architecture")
print("5. Use 8-bit quantization for deployment")
print("6. Use TensorFlow Lite for mobile/edge devices")
```
